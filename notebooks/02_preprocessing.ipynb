# 02 — Preprocessing
Demonstrates the cleaning steps and resulting feature matrices.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path('..') / 'src'))

from preprocess import (
    load_and_clean,
    build_song_features,
    build_user_features,
    build_joint_features,
)

df = load_and_clean()
print('Cleaned shape:', df.shape)
df[['user_id', 'gender', 'hometown', 'genre', 'song_lang_list', 'release_year']].head(8)
```

## Genre cleaning check

```python
print('Unique genres after cleaning:')
print(sorted(df['genre'].unique()))
```

## Song language cleaning check

```python
from collections import Counter
all_langs = Counter(l for lst in df['song_lang_list'] for l in lst)
print('Song languages after cleaning:')
for lang, cnt in all_langs.most_common():
    print(f'  {lang}: {cnt}')
```

## Song feature matrix

```python
sf = build_song_features(df)
print('Song feature matrix shape:', sf.shape)
print('Columns:', list(sf.columns))
sf.head()
```

## User feature matrix

```python
uf = build_user_features(df)
print('User feature matrix shape:', uf.shape)
print('Columns:', list(uf.columns))
uf.head()
```

## Joint feature matrix (used by K-means)

```python
jf = build_joint_features(df)
print('Joint feature matrix shape:', jf.shape)

# Verify no NaNs
assert jf.isna().sum().sum() == 0, 'Found NaNs!'
print('No NaN values — preprocessing complete.')
```
